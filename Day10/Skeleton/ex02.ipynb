From "../Data/sample.html"

```python
file_name = "../../Data/sample.html"
with open(file_name, 'r') as file:
    html_data = file.read()
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_data, "html.parser")
print(soup)
```

- Return the title of the html content

```python
soup.find("title")
```

- Return all `<h2>`

```python
soup.find_all("h2")
```

- Return all where its id is email

```python
soup.select("#email")  # '#' shorthand for id
```

```python
soup.select("[id='email']")
```

- Return all `<a>` where its class is "resource"

```python
soup.select("a.resource")  # '.' shorthand for class
```

```python
soup.select("a[class='resource']")
```

- Return all`<a>` where its class is resource and href is 'https://docs.python.org'

```python
soup.select("a[class='resource'][href='https://docs.python.org']")
```

```python
soup.select("li a.resource[href='https://docs.python.org']")
```

- Return the NaviableString (string val) of elements, which tag is an anchor `<a>` in an array

```python
output = []
for anchor in soup.select("a"):  # find_all would also work here, but get in the habit of using select
    output.append(anchor.string)
output
```
