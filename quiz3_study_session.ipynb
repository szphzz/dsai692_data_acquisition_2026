## Day 10 - BeautifulSoup

### Create a soup object using HTML documents - from the example quiz

```python
import requests
from bs4 import BeautifulSoup

def retrieve_ul_children(url):
    html_text = requests.get(url).content
    soup = BeautifulSoup(html_text, "html.parser")
    output = []
    # for li in soup.select("ul li"):
    # for li in soup.select("[id='1']"):
    # for li in soup.select(".animal[id='1']"):
    for li in soup.select("ul .fruit[id='1']"):
        output.append(li.string)
    return output
```

```python
def traverse(url):
    html_text = requests.get(url).content
    soup = BeautifulSoup(html_text, "html.parser")
    output = []
    for fruit in soup.select("ul"):
        # descendants is a list of newlines, list items, and Navigable strings
        for item in list(fruit.descendants):  # need to convert generator to list
            if item != "\n":
                output.append(item)
    return output
```

#### Finding and Accessing Objects
- .find()
    - returns text
    - can use tag name
    - can NOT use attributes
- .find_all()
    - returns list
    - can use tag name
    - can NOT use attributes
- .select()
    - returns list
    - can use tag name
    - can use attributes

- attributes go after the tag 
    - ex: \<li id="1" class="fruit"\>

- \<ul\> is unordered list (bullet points)
- \<ol\> is ordered list (numbers)
- \<li\> is list item

- we won't be asked to write traversing code, just know the concepts

## Day 12 - Playwright

### From the example quiz

```python
from playwright.sync_api import sync_playwright

def fill_and_click(url, text):
    with sync_playwright() as p:  # context management, closes browser automatically
        browser = p.chromium.launch(headless=False)
        # creating a context is optional
        page = browser.new_page()
        page.goto(url)
        # page.locator("textarea[name='q']").fill(text)
        page.locator("textarea[name='q']").click()  # would do the same as above
        page.keyboard.type(text)  # .press() wouldn't work here because it's only for one key
        page.locator("button").click()
        # page.screenshot(path="output.png")  # won't be tested
        browser.close()
```
