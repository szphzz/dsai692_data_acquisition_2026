Ex 4. Return a list of dictionary of `{page_number: text}` all the pages including `anchovies`

```python
from pypdf import PdfReader

url = 'https://data.cnra.ca.gov/dataset/78ac95ac-9665-4e53-aea6-ad0762055cce/resource/7247f0f6-915f-47fb-a9a2-81789cf40b37/download/microplastic-levels-in-sf-bay-final-report.pdf'
key_word = 'anchovies'

response = requests.get(url)  # retrieve the data
data = BytesIO(response.content)  # read in as bytes
pdf = PdfReader(data)  # create a PDF object

output = []
for page_no, page in enumerate(pdf.pages, 1):  # loop through pages
    page_text = page.extract_text()  # to get text
    if "anchovies" in page_text.lower():  # case-insensitive
        output.append({page_no: page_text})
```
